```python
# Allocation Matrix

allocation = [
    [0, 1, 0],  # P0
    [2, 0, 0],  # P1
    [3, 0, 2],  # P2
    [2, 1, 1],  # P3
    [0, 0, 2]   # P4
]


# Maximum Matrix

maximum = [
    [7, 5, 3],  # P0
    [3, 2, 2],  # P1
    [9, 0, 2],  # P2
    [2, 2, 2],  # P3
    [4, 3, 3]   # P4
]


# Available Resources

available = [3, 3, 2]
```

```python
Need = []

for i in range(len(allocation)):

    each_process_need = []

    for j in range(len(available)):

        value = maximum[i][j] - allocation[i][j]

        each_process_need.append(value)

    Need.append(each_process_need)


print("Need Matrix")

for i in range(len(Need)):
    print("P", i, ":", Need[i])
```

```python
Work = available.copy()

Finish = []

for i in range(len(allocation)):
    Finish.append(False)

print("Work =", Work)
print("Finish =", Finish)
```

```python
safe_sequence = []

while True:

    found = False

    for i in range(len(allocation)):

        if Finish[i] == False:

            for j in range(len(Work)):

                if Need[i][j] > Work[j]:
                    break

            else:

                for j in range(len(Work)):
                    Work[j] = Work[j] + allocation[i][j]

                Finish[i] = True

                safe_sequence.append(i)

                print("P", i, "finished")

                found = True

    if found == False:
        break


print("Safe Sequence:", safe_sequence)
```
